```python
import mediapipe as mp
import numpy as np
import os
import cv2 as cv
```

```python
INPUTROOT = "G:/My Drive/school/Uni/PHD/Project_with_Sue/Dynamite"
input_video_file_name = "evan_neck.mov"
output_video_file_name = "evan_neck_output.mov"

file_name = os.path.join(INPUTROOT, input_video_file_name)
out_file_name = os.path.join(INPUTROOT, output_video_file_name)
```

```python
# use mediapipe to obtain the landmarks at every frame
mp_drawing = mp.solutions.drawing_utils
mp_pose = mp.solutions.pose
cap = cv.VideoCapture(file_name)
# height and width of video
height = int(cap.get(cv.CAP_PROP_FRAME_HEIGHT))
width = int(cap.get(cv.CAP_PROP_FRAME_WIDTH))

# open a stream to store drawn landmarks

out = cv.VideoWriter(out_file_name, cv.VideoWriter_fourcc(*'mp4v'), 30, (width, height))
```

```python
raw_landmarks = []

with mp_pose.Pose(min_detection_confidence=0.5, min_tracking_confidence=0.5) as pose:
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        image = cv.cvtColor(frame, cv.COLOR_BGR2RGB)
        image.flags.writeable = False

        results = pose.process(image)

        image.flags.writeable = True
        image = cv.cvtColor(image, cv.COLOR_RGB2BGR)

        if results.pose_landmarks:
            mp_drawing.draw_landmarks(image, results.pose_landmarks, mp_pose.POSE_CONNECTIONS)
            raw_landmarks.append(results.pose_landmarks)
            # save the drawn frame
            out.write(image)
out.release()
    
```
